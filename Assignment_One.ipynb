# Assignment 1: Mini DSA Project — Student Performance Analyzer
## 1. Problem Explanation

The Student Performance Analyzer helps a lecturer manage and analyse student marks. It allows the lecturer to enter and display marks, identify the highest and lowest marks, search for specific marks, and sort marks in ascending order.

The program uses a Python list to store marks and implements Linear Search and Bubble Sort to demonstrate fundamental data structures and algorithms.


## 2. Algorithm Design

### High-Level Program Flow
1. Initialize an empty list `marks = []`.
2. Display a menu with the following choices:
   - 1: Input student marks
   - 2: Display marks
   - 3: Find highest and lowest marks
   - 4: Search for a specific mark
   - 5: Sort marks in ascending order
   - 6: Exit
3. Read user choice and execute the corresponding procedure until the user chooses to exit.


### Step-by-Step Algorithms

#### A. Accepting Marks
1. Prompt the lecturer for the number of marks $N$.
2. For each input $i$ from 1 to $N$:
   a. Validate that the mark is a numeric value between 0 and 100.
   b. Append the mark to `marks`.

#### B. Finding Highest and Lowest Marks
1. If `marks` is empty, notify the user.
2. Initialize `highest = marks[0]` and `lowest = marks[0]`.
3. Loop through `marks` from index 1 to $N-1$:
   a. If `mark > highest`, update `highest = mark`.
   b. If `mark < lowest`, update `lowest = mark`.
4. Return `highest` and `lowest`.

#### C. Searching Algorithm (Linear Search)
1. Accept `target` mark to search.
2. Set `found_index = -1`.
3. Loop through `marks` with index $i$ from 0 to $N-1$:
   a. If `marks[i] == target`:
      i. Set `found_index = i`.
      ii. Break loop.
4. If `found_index != -1`, return the position. Otherwise, return not found.

#### D. Sorting Algorithm (Bubble Sort)
1. Let $N$ be the length of `marks`.
2. Create a shallow copy of `marks` so original order is preserved if needed.
3. For $i$ from 0 to $N-1$:
   a. Set `swapped = False`.
   b. For $j$ from 0 to $N - i - 2$:
      i. If `marks[j] > marks[j+1]`:
         - Swap `marks[j]` and `marks[j+1]`.
         - Set `swapped = True`.
   c. If `swapped == False`, break early(array already sorted).
4. Return the sorted list.

```python
"""
Student Performance Analyzer
Implements custom Linear Search, Bubble Sort, Min/Max finding, and an interactive menu.
"""

def input_marks():
    """Prompts lecturer to enter student marks with basic range validation."""
    marks = []
    try:
        count = int(input("Enter number of students: "))
        if count <= 0:
            print("Please enter a positive integer.")
            return marks
        
        for i in range(count):
            while True:
                try:
                    val = float(input(f"Enter mark for student {i + 1} (0 - 100): "))
                    if 0 <= val <= 100:
                        marks.append(val)
                        break
                    else:
                        print("Mark must be between 0 and 100. Try again.")
                except ValueError:
                    print("Invalid input. Please enter a valid number.")
        print(f"\nSuccessfully recorded {len(marks)} marks.")
    except ValueError:
        print("Invalid number of students.")
    return marks


def display_marks(marks):
    """Displays current collection of marks."""
    if not marks:
        print("No marks currently recorded.")
        return
    print("\n--- Current Student Marks ---")
    for idx, mark in enumerate(marks, start=1):
        print(f"Student {idx}: {mark}")


def find_highest_lowest(marks):
    """Determines highest and lowest marks without built-in max()/min()."""
    if not marks:
        print("No marks recorded to evaluate.")
        return None, None

    highest = marks[0]
    lowest = marks[0]

    for mark in marks[1:]:
        if mark > highest:
            highest = mark
        if mark < lowest:
            lowest = mark

    return highest, lowest


def linear_search(marks, target):
    """
    Searches for a mark using Linear Search.
    Returns a list of all 0-based indices where the target occurs.
    """
    matching_indices = []
    for i in range(len(marks)):
        if marks[i] == target:
            matching_indices.append(i)
    return matching_indices


def bubble_sort(marks):
    """
    Sorts a list in ascending order using Bubble Sort without sort() or sorted().
    Operates on a copy to prevent unintended in-place mutation of raw input.
    """
    sorted_list = marks.copy()
    n = len(sorted_list)

    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if sorted_list[j] > sorted_list[j + 1]:
                # Swap elements
                sorted_list[j], sorted_list[j + 1] = sorted_list[j + 1], sorted_list[j]
                swapped = True
        # Optimization: break if already sorted
        if not swapped:
            break

    return sorted_list


def main_menu():
    """Provides a console menu for the lecturer to execute operations."""
    marks = []

    while True:
        print("\n========================================")
        print("   STUDENT PERFORMANCE ANALYZER MENU   ")
        print("========================================")
        print("1. Enter Student Marks")
        print("2. Display Marks")
        print("3. View Highest and Lowest Marks")
        print("4. Search for a Mark")
        print("5. Sort Marks (Ascending Order)")
        print("6. Exit")
        print("========================================")

        choice = input("Enter choice (1-6): ").strip()

        if choice == '1':
            marks = input_marks()

        elif choice == '2':
            display_marks(marks)

        elif choice == '3':
            if not marks:
                print("Please enter marks first (Option 1).")
            else:
                highest, lowest = find_highest_lowest(marks)
                print(f"\nHighest Mark: {highest}")
                print(f"Lowest Mark:  {lowest}")

        elif choice == '4':
            if not marks:
                print("Please enter marks first (Option 1).")
            else:
                try:
                    target = float(input("Enter mark to search for: "))
                    results = linear_search(marks, target)
                    if results:
                        print(f"\nMark {target} was FOUND at index/indices: {results} (Student #{[r + 1 for r in results]})")
                    else:
                        print(f"\nMark {target} was NOT found in the dataset.")
                except ValueError:
                    print("Invalid input. Please enter a numerical mark.")

        elif choice == '5':
            if not marks:
                print("Please enter marks first (Option 1).")
            else:
                sorted_marks = bubble_sort(marks)
                print("\n--- Marks Sorted in Ascending Order ---")
                print(sorted_marks)

        elif choice == '6':
            print("Exiting Student Performance Analyzer. Goodbye!")
            break

        else:
            print("Invalid choice. Please choose an option between 1 and 6.")

# Run the program
if __name__ == "__main__":
    main_menu()
```

## 3. Data Structures and Algorithms Analysis

### 3.1 Data Structure: Python List

A Python list stores student marks and supports dynamic sizing, direct index access in `O(1)` time, and element swapping.

### 3.2 Searching Algorithm: Linear Search

Linear Search checks each mark against the target and records matching indices. It works on unsorted data.

- **Time complexity:** `O(N)`

### 3.3 Sorting Algorithm: Bubble Sort

Bubble Sort compares adjacent elements and swaps them when they are out of order. Each pass moves the largest unsorted element towards its correct position. A `swapped` flag stops the algorithm early if no swaps occur.

### 3.4 Complexity Analysis

| Algorithm | Best Case | Average Case | Worst Case | Auxiliary Space |
|---|---|---|---|---|
| Find highest/lowest | `O(N)` | `O(N)` | `O(N)` | `O(1)` |
| Linear Search (all matches) | `O(N)` | `O(N)` | `O(N)` | `O(N)` worst case for stored indices |
| Bubble Sort | `O(N)` | `O(N²)` | `O(N²)` | `O(N)` including the copied list |

`O(N²)` means that the number of operations can grow proportionally to the square of the input size. For example, doubling the number of marks can approximately quadruple the work in the average and worst cases of Bubble Sort.

Bubble Sort's best case is `O(N)` because the `swapped` flag allows it to stop after one pass when the list is already sorted.

```python


# Test dataset containing 10 student marks
test_marks = [78.5, 45.0, 92.0, 63.5, 88.0,
              55.0, 92.0, 34.0, 71.5, 80.0]

print("=" * 50)
print("TEST DATASET (10 Student Marks):")
print(test_marks)
print("=" * 50)

# Operation 1: Find the highest and lowest marks
high, low = find_highest_lowest(test_marks)
print("\n[Operation 1 - Highest and Lowest]")
print(f"Highest Mark: {high}")
print(f"Lowest Mark: {low}")

# Operation 2: Search for existing and missing marks
print("\n[Operation 2 - Linear Search]")
for target in [92.0, 100.0]:
    result = linear_search(test_marks, target)
    if result:
        print(f"{target} found at indices: {result}")
    else:
        print(f"{target} not found.")

# Operation 3: Sort the marks without changing the test data
print("\n[Operation 3 - Bubble Sort]")
sorted_test_marks = bubble_sort(test_marks.copy())
print(f"Original Marks: {test_marks}")
print(f"Sorted Marks:   {sorted_test_marks}")
```
